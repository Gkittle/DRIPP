```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
plt.figure(figsize=(10,10))
#list_fail = [0,1,17]
list_fail = []
final_cost = []
for i in range(50):
    if i in list_fail:
        continue
    list = []
    with open(f"/home/gkittles/SB_DRIPP_Results/5_20_25_Parameter_Sweep/slurm-65730591_{i+1}.out") as f:
        for line in f:
            ilist = [elt.strip() for elt in line.split('; ')]
            list.append(ilist)
    nfe = []
    cost = []
    for val in list:
        nfe.append(int(val[0].split(" ")[1]))
        cost.append(float(val[2].split(" ")[0]))
    color = plt.cm.cool(i/50)
    plt.plot(nfe,cost, label = f"{i}", color=color);
    final_cost.append(cost[-1])
#plt.legend();
plt.xlabel('NFE');
plt.ylabel('System Cost');
```

```python
plt.bar(range(50), final_cost);
```
